```python
import matplotlib.pyplot as plt

marks = []
# getting input for 5 students
for i in range(5):
    m = float(input("Enter marks for student: "))
    marks.append(m)

total = sum(marks)
avg = total / 5
highest = max(marks)
lowest = min(marks)

print("Total marks:", total)
print("Average marks:", avg)
print("Highest mark:", highest)
print("Lowest mark:", lowest)

students = ['S1', 'S2', 'S3', 'S4', 'S5']
plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN']
sales = [120, 150, 180, 170, 200, 220]

avg_sales = sum(sales) / len(sales)
print("Average monthly sales:", avg_sales)

# finding the highest month
max_sale = max(sales)
max_index = sales.index(max_sale)
best_month = months[max_index]
print("Month with highest sales:", best_month)

# Line Chart
plt.plot(months, sales, marker='o')
plt.title("Monthly Sales (Line)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# Bar Chart
plt.bar(months, sales)
plt.title("Monthly Sales (Bar)")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
import matplotlib.pyplot as plt

height = [150, 155, 160, 165, 170, 175]
weight = [45, 50, 54, 60, 67, 50]

avg_h = sum(height) / len(height)
avg_w = sum(weight) / len(weight)

print("Average height:", avg_h)
print("Average weight:", avg_w)

plt.scatter(height, weight)
plt.title("Height vs Weight")
plt.xlabel("Height (cm)")
plt.ylabel("Weight (kg)")
plt.show()
```

```python
import matplotlib.pyplot as plt

clubs = {
    'Programming': 45,
    'Robotics': 50,
    'Cyber Security': 54,
    'Web Programming': 60
}

total_students = sum(clubs.values())
print("Total number of students:", total_students)

for club, count in clubs.items():
    percent = (count / total_students) * 100
    print(club, "percentage:", percent)

names = list(clubs.keys())
counts = list(clubs.values())

plt.pie(counts, labels=names, autopct='%1.1f%%')
plt.title("Students in Clubs")
plt.show()
```

```python
workshop_a = {'Alice', 'Bob', 'Charlie', 'David'}
workshop_b = {'Charlie', 'David', 'Eva', 'Frank'}

both = workshop_a.intersection(workshop_b)
print("Attended both:", both)

either = workshop_a.union(workshop_b)
print("Attended either:", either)

only_a = workshop_a.difference(workshop_b)
print("Only Workshop A:", only_a)

total_unique = len(either)
print("Total unique students:", total_unique)
```
