Line Plot

```python
import matplotlib.pyplot as plt
Subjects=['Maths','Physics','Chemistry','English','Biology']
Marks=[60,34,20,70,90]
plt.plot(Subjects,Marks,marker=('o'))
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Marks in Subjects')
plt.show()
```

```python
import matplotlib.pyplot as plt
Subjects1=[]
Marks1=[]
n=int(input('How many number do you Subjects:'))
for i in range(n):
    Subject=input(f'Enter the Subject{i+1}:')
    Mark=int(input(f'Enter the Marks of {Subject}:'))
    Subjects1.append(Subject)
    Marks1.append(Mark)

    
plt.plot(Subjects1,Marks1,marker=('o'))
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Marks in Subjects')
plt.show()
    
```

BarChart

```python
import matplotlib.pyplot as plt
Subjects=['Maths','Physics','Chemistry','English','Biology']
Marks=[60,34,20,70,90]
plt.bar(Subjects,Marks)
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.title('Bar Chart of Marks in Subjects')
plt.show()
```
