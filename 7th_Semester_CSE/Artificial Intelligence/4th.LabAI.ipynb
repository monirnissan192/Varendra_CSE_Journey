```python
from collections import deque
Graph = {
    's':['A','B','c'],
    'A':['D','E'],
     'B':['G'],
     'C':['F'],
     'D':['H'],
     'E':['G'],
     'F':['G'],
     'G':[],
     'H':[]    
}
Start='S'
Goal='G'
queue=deque()
visited=[]
```

```python
def BFS(Start,Goal):
    queue.append(Start)
    while queue:
        node=queue.popleft()

        if node not in visited:
            print(node)
            visited.append(node)
            if node == Goal:
                return '\n Goal Satte Found !'
            for i in Graph[node]:
                queue.append(i)
    return'\nGoal State not Found !'            
```

```python
BFS(Start,Goal)
```
