This is Uniform Search code, here in this code I solve BFS(Breadth First Search)

```python
from collections import deque

Graph= {
    'S':['A','B','C'],
    'A':['D','E'],
    'B':['G'],
    'C':['F'],
    'D':['E','H'],
    'E':['G'],
    'F':['G'],
    'G':[],
    'H':[]
}
Start='S'
Goal='G'
queue=deque()
Visited=[]
#In Jupyter, variables stay in memory between cell runs. 
# We must reset 'Visited' and 'queue' to empty each time so the function doesn't use old data.
#First time output same but 2nd time output can be change, But BFS Algorithm implementation is correct
```

```python
def BFS(Start,Goal):
    queue.append(Start)
    while queue:
        node=queue.popleft()

        if node not in Visited:
            print(node)
            Visited.append(node)
            if node == Goal:
                return '\n Goal State is Found !'
            for i in Graph[node]:
                queue.append(i)
    return '\n Goal State is Not Found !'            
```

```python
Result = BFS(Start,Goal)
```

```python
print(Result)
```
