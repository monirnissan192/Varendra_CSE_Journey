```python
for i in Course:
    print(i)
```

In Set we cannot access any data from Set here basically that is unordered They do not supporting indexing like[course[0]]. For We convert the set into list

```python
Course_list = list(Course)

# 2. Print the new list to see the order
print("The list is:", Course_list)

# 3. Now you can access data directly using an index!
print("The first course is:", Course_list[0])
print("The second course is:", Course_list[1])
```

Here we using Queue. We use Dequeue,dequeue means double ended queue...... 

```python
from collections import deque
Numbers=deque()
#For adding elements we use .append()method
Numbers.append(1)
Numbers.append(2)
Numbers.append(3)
```

```python
Numbers.pop()
#For Remove elements from the right side we use .pop() method
Numbers.popleft()
#For Remove elements from the left side we use .popleft() method
```

```python
#Dictionary declaration
Club_Info = {
    'Programing':50,
    'Robotics':44,
    'CyberSecurity':30,
    'Web_Progaraming':60
}
```

```python
Total_Student=sum(Club_Info.values())
Total_Student
```

```python
Student_Percentage=[i/(Total_Student)*100 for i in Club_Info.values()]
Student_Percentage
```

Same Output,but easy way

```python
Ptr_Student=[]
for x in Club_Info.values():
    Ptr_Student.append((x/Total_Student)*100)

Ptr_Student
```

```python
import matplotlib.pyplot as plt

# Updated the title to match the graph type
plt.title('Club Membership Pie Chart')

# Added %% to the autopct to show the percentage symbol
plt.pie(Ptr_Student, labels=Club_Info.keys(), autopct='%.2f%%')

# Command to display the graph
plt.show()
```
