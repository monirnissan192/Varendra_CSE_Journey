```python
from collections import deque
Graph= {
    'S':['A','B','C'],
    'A':['D','E'],
    'B':['G'],
    'C':['F'],
    'D':['E','H'],
    'E':['G'],
    'F':['G'],
    'G':[],
    'H':[]
}
Start='S'
Goal='G'
```

```python
def BFS(Start,Goal):
    queue=deque()
    Visited=[]
    queue.append(Start)
    while queue:
        node=queue.popleft()

        if node not in Visited:
            print(node)
            Visited.append(node)
            if node== Goal:
                return '\n Goal State Found !'
            for i in Graph[node]:
               queue.append(i)
    return'\n Goal State Not Found !'
```

```python
Result=BFS(Start,Goal)
```

```python
print(Result)
```

```python
Graph= {
    'A':['C','B'],
    'B':['E','D'],
    'C':['F'],
    'D':['G'],
    'E':['G'],
    'F':['H'],
    'G':['H'],
    'H':[]
}
Start='A'
Goal='H'
```

```python
def DFS(Start,Goal):
    Stack=[]
    Visited=[]
    Stack.append(Start)
    while Stack:
        node=Stack.pop()
        if node not in Visited:
            print(node)
            Visited.append(node)

            if node == Goal:
                return '\n Goal State Found !'

            for i in Graph[node]:
                Stack.append(i)
    return '\n  Goal State Not Found!'
```

```python
Result=DFS(Start,Goal)
```

```python
print(Result)
```
