Deapth First Search

```python

Graph= {
    'A':['C','B'],
    'B':['E','D'],
    'C':['F'],
    'D':['G'],
    'E':['G'],
    'F':['H'],
    'G':['H'],
    'H':[]
}
Start='A'
Goal='H'
```

```python
def DFS(Start,Goal):
    
    stack=[]
    Visited=[]
    stack.append(Start)
    while stack:
        node=stack.pop()

        if node not in Visited:
            Visited.append(node)
            print(node)
            if node == Goal:
                return '\n Goal State is Found !'
            for i in Graph[node]:
                stack.append(i)
    return '\n Goal State is Not Found !' 
```

```python
Result = DFS(Start,Goal)
```
