Load the Iris dataset into a DataFrame. 

```python
import matplotlib.pyplot as plt
import seaborn as mn
import pandas as pd
```

```python
mn.get_dataset_names()
```

```python
df= mn.load_dataset('iris')
```

```python
df.head()
```

```python
df.describe()
```

```python
species_data=df['species'].value_counts()
print(species_data)
species_name=species_data.index
species_count=species_data.values
```

Visualize the class distribution of the three species using a bar chart.  

```python
plt.bar(species_data,species_count)
plt.xlabel(species_name)
plt.ylabel(species_count)
plt.show
```

Find mean, median, mode, and standard deviation of a feature

```python
df['petal_length'].median()
df['petal_length'].mean()
df['petal_length'].mode()
df['petal_length'].std()
```

Create a histogram to visualize the distribution of the sepal_length feature

```python
plt.hist(df['sepal_length'],[3.5,4.5,5.5,6.5,7.5],edgecolor='black')
plt.xlabel('sepal_length')
plt.ylabel('Frequency')
plt.show()
```

Create a scatter plot to visualize the relationship between sepal_length and petal_length, using different colors for each species

```python
mn.scatterplot (data=df, x='sepal_length', y='petal_length')
plt.title('Sepal Length vs. Petal Length')
plt.show()
```

Create a box plot of the sepal_length feature with individual data points to identify potential outliers  


```python
mn.boxplot(data=df, x='species', y='sepal_length',color='white',showfliers=False)
mn.stripplot(data=df, x='species', y='sepal_length',alpha=0.6,jitter=0.2)
plt.show()
```
